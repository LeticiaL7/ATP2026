#### a)

```python
def strCount(s, subs):
    cont = 0

    for i in range(len(s) - len(subs) + 1):
        if s[i:i+len(subs)] == subs:
            cont += 1

    return cont

strCount("catcowcat", "cat") # --> 2
strCount("catcowcat", "cow") # --> 1
strCount("catcowcat", "dog") # --> 0
```

#### c)

```python
def reduxInt(n):
    while n >= 10:
        soma = 0

        for digito in str(n):
            soma += int(digito)

        n = soma

    return n
```

```python
def myIndexOf(s1, s2):
    for i in range(len(s1) - len(s2) + 1):
        if s1[i:i+len(s2)] == s2:
            return i
    return -1
```

#### d)

```python
def insPost(redeSocial, conteudo, autor, dataCriacao, comentarios):

    i = 0

    for post in redeSocial:
        num = int(post['id'][1:])   # tira o 'p'

        if num > i:
            i = num

    novoPost = {
        'id': 'p' + str(i + 1),
        'conteudo': conteudo,
        'autor': autor,
        'dataCriacao': dataCriacao,
        'comentarios': comentarios
    }

    redeSocial.append(novoPost)

    return redeSocial
```

#### e)

```python
def remPost(redeSocial, id):
   for post in redeSocial:
        if post['id'] == id:
            redeSocial.remove(post)

   return redeSocial 
```

#### f)

```python
def postsPorAutor(redeSocial):
    dist = {}

    for post in redeSocial:
        autor = post['autor']

        if autor in dist:
            dist[autor] += 1
        else:
            dist[autor] = 1

    return dist
```

#### g)

```python
def comentadoPor(redeSocial, autor):

    res = []

    for post in redeSocial:
        autores = []

        for comentario in post['comentarios']:
            autores.append(comentario['autor'])

        if autor in autores:
            res.append(post)

    return res
```
